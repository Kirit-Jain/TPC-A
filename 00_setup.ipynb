```python
%sql
Create catalog if not exists tpc_a_dev;

Create schema if not exists tpc_a_dev.bronze;
Create schema if not exists tpc_a_dev.silver;
Create schema if not exists tpc_a_dev.gold;

Create volume if not exists tpc_a_dev.bronze.raw;
Create volume if not exists tpc_a_dev.bronze.pipeline_state;
```

```python
%sql
-- Verify all three schemas exist in tpc_a_dev
SHOW SCHEMAS IN tpc_a_dev;

-- Confirm bronze, silver, gold are present (excluding information_schema and default)
SELECT catalog_name, schema_name
FROM tpc_a_dev.information_schema.schemata
WHERE schema_name IN ('bronze', 'silver', 'gold')
ORDER BY schema_name;
```
